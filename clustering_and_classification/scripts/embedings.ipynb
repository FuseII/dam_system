BERT / Multilingual BERT (mBERT)

Плюсы:<br>
✅ Лучшее качество для контекстных задач (NER, классификация).<br>
✅ Учитывает многозначность слов.<br>

Минусы:<br>
❌ Требует GPU для эффективной работы.<br>
❌ Сложная токенизация (WordPiece).<br>
В BERT (и Multilingual BERT) модель выводит не один вектор на слово, а тензор размерности [Batch_size, Sequence_length, Hidden_size], потому что:

   ## 1. Контекстная природа BERT
Каждый токен получает уникальный эмбеддинг, зависящий от окружения (контекста).

Например, слово "банк" в предложениях:

"Я положил деньги в банк" → (финансовое учреждение).

"Мы сидели на берегу банка" → (берег реки).
BERT выведет разные вектора для одного и того же слова в разных контекстах.

   ## 2. Структура вывода BERT
Для входного текста "Привет мир" (разбитого на токены [CLS], "Привет", "мир", [SEP]):

Модель возвращает тензор размерности [1, 4, 768] (где 768 — размерность эмбеддинга в bert-base).

1 — размер батча.

4 — количество токенов (включая служебные [CLS] и [SEP]).

768 — скрытый размер модели.
## 3. Как получить один вектор?
Если нужен один эмбеддинг для всего предложения, используют:<br>
Пул-методы (усреднение или суммирование векторов токенов).<br>
Вектор [CLS] — специальный токен, который (теоретически) содержит информацию о всем предложении.

```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")
model = AutoModel.from_pretrained("bert-base-multilingual-cased")

# Токенизация и получение эмбеддингов
inputs = tokenizer("Привет, мир!", return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)

# Усреднение эмбеддингов токенов
sentence_embedding = torch.mean(outputs.last_hidden_state[:, 1:-1, :], dim=1)

sentence_embedding 
```
