# Обучение модели на новых сущностях

```python
text = "Часть территории национального парка 'Лосиный остров', расположенная в границах города Москвы  ППМ № 769-ПП ORG от 11.04.2025 ЗХ-1 Зона хозяйственного назначения".lower()
sub_string = "Лосиный остров".lower()
left_pos = text.find(sub_string)
right_pos = left_pos + len(sub_string)

print(f"{left_pos} {right_pos} {text[left_pos:right_pos]}")
```

```python
# Пример аннотированных данных
annotated_examples = [

    ("Часть территории национального парка 'Лосиный остров', расположенная в границах города Москвы  ППМ № 769-ПП ORG от 11.04.2025 ЗХ-1 Зона хозяйственного назначения",
     {"entities":[(38, 52, "LOC")]})
]
```

```python
# Получаем NER компонент
ner = nlp.get_pipe("ner")
```

```python
# Добавляем новые labels
new_labels = set()
for examples in annotated_examples:
    for ent in examples[1].get("entities", []):
        print(ent[2])
        new_labels.add(ent[2])
    
for label in new_labels:
    if label not in ner.labels:
        ner.add_label(label)
        print(f"Добавлен новый label: {label}")

# Обучение только на новых данных
# отключаем все остальные pipe
other_pipes = [pipe for pipe in nlp.pipe_names if pipe != "ner"]
with nlp.disable_pipes(*other_pipes):
    optimizer = nlp.resume_training()
# 
iteration_number = 30
for iteration in range(iteration_number):
            random.shuffle(annotated_examples)
            losses = {}

            for example in annotated_examples:
                text = example[0]
                annotations = example[1]
                doc = nlp.make_doc(text)
                train_example = Example.from_dict(doc, annotations)
                nlp.update([train_example], sgd=optimizer, losses=losses, drop=0.4)

            if iteration % 5 == 0:
                print(f"Iteration {iteration}, Loss: {losses['ner']}")
```
